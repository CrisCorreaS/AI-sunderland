```python
import nltk

nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string

# Texts
text1 = "Fauvism, led by Henri Matisse, is known for its bold use of color and emphasis on emotional expression over realism."
text2 = "Joaquín Sorolla, a Spanish painter, was renowned for his masterful use of light and vibrant seascapes, often depicting beach scenes."
text3 = "The Renaissance, spanning from the 14th to the 17th century, marked a rebirth of classical art, with artists like Leonardo da Vinci and Michelangelo revolutionizing painting and sculpture."

art_history_facts = [text1, text2, text3]
processed_facts = []

for fact in art_history_facts: 
    fact = fact.lower()

    fact = fact.translate(str.maketrans("", "", string.punctuation))

    tokens = word_tokenize(fact)

    stop_words = set(stopwords.words("english"))
    filtered_tokens = [word for word in tokens if word not in stop_words]

    processed_facts.append(filtered_tokens)

print("After Text Preprocessing: ")
for i, processed in enumerate(processed_facts):
    print(f"Text {i + 1}: {processed}")
```
