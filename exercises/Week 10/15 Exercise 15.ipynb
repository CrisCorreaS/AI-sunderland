Using the previous examples, develop a text summarisation script that would
read a text file and summarise it into 3 sentences.
The steps to do so are:
1. Tokenize the sentences and words in the text.
2. Remove any stopwords from the text.
3. Calculate the frequencies of each word, excluding the stopwords.
4. Calculate the score for each sentence based on the frequencies of the words it
contains, considering only sentences with less than 30 words.
5. Select the top 2 sentences with the highest scores to generate a summary.
6. Combine these top sentences to create the final summary

```python
import nltk
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from collections import defaultdict

nltk.download("punkt")
nltk.download("stopwords")


text = "Frida Kahlo was a Mexican painter known for her vibrant and deeply personal artwork, which often explored themes of identity, the human body, and death. Her art is characterized by its unflinching portrayal of her own experiences, including the physical and emotional pain she endured throughout her life. Kahlo’s paintings are predominantly self-portraits, reflecting her intense introspection and the various stages of her life. She used a mix of styles, drawing inspiration from indigenous Mexican traditions, European influences such as Realism, Symbolism, and Surrealism, and her own Catholic heritage. Her choice of small-scale oil paintings on metal, inspired by naive devotional paintings, was a unique feature of her work. Some of her notable paintings include “Frieda and Diego Rivera” (1931) and “The Two Fridas” (1939). These works not only depict her physical appearance but also her inner emotional states and her complex relationship with her husband, the muralist Diego Rivera. Her art often includes symbolic elements that represent her struggles with physical pain, her Mexican identity, and her connection to nature. Kahlo’s work gained significant international recognition in the 1970s and continues to be celebrated for its emotional depth and cultural significance. Her paintings have been featured in numerous retrospectives and exhibitions worldwide, including at the Tate Modern in London and the Louvre in Paris, where “The Frame” (c. 1938) became the first work by a 20th-century Mexican artist to be included in the museum’s collection."

# A. Tokenize the sentences and words in the text.
sentences = sent_tokenize(text)

# B. Remove any stopwords from the text.
text = text.lower()  
text = text.translate(str.maketrans("", "", string.punctuation))  
words = word_tokenize(text)  

stop_words = set(stopwords.words("english"))
filtered_words = [word for word in words if word not in stop_words]  

# C. Calculate the frequencies of each word, excluding the stopwords.
word_frequencies = defaultdict(int)
for word in filtered_words:
    word_frequencies[word] += 1  

# D. Calculate the score for each sentence based on the frequencies of the words it contains, considering only sentences with less than 30 words.
sentence_scores = defaultdict(int)
for sentence in sentences:
    sentence_word_count = word_tokenize(sentence)  
    sentence_length = len(
        sentence_word_count
    )  

    if sentence_length < 30:  
        for word in sentence_word_count:
            if word.lower() in word_frequencies:
                sentence_scores[sentence] += word_frequencies[
                    word.lower()
                ]  

# E. Select the top 2 sentences with the highest scores to generate a summary.
sorted_sentences = sorted(
    sentence_scores.keys(), key=lambda x: sentence_scores[x], reverse=True
)  

num_sentences_to_select = min(
    3, len(sorted_sentences)
)  
summary_sentences = sorted_sentences[
    :num_sentences_to_select
]  

# F. Combine these top sentences to create the final summary
summary = " ".join(summary_sentences)

print("Summary:")
print(summary)
```
