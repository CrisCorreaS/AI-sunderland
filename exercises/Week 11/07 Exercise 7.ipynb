Now experiment with different roles by changing the following part,
and compare the results, so for example as it to explain it as a computer science
expert vs a 5 year old kid and so on.


```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer

model_name = "Qwen/Qwen2.5-0.5B-Instruct"
model = AutoModelForCausalLM.from_pretrained(
    model_name, torch_dtype="auto", device_map="auto"
)
tokenizer = AutoTokenizer.from_pretrained(model_name)

prompt = "Give me a short introduction to large language models."
```

# Experiment 1
Qwen is a computer science expert

```python
messages = [
    {
        "role": "system",
        "content": "You are a computer science expert.",
    },
    {"role": "user", "content": prompt},
]

input_text = ""
for message in messages:
    input_text += (
        f"{message['role']}: {message['content']}\n"  # Concatenate message content
    )

model_inputs = tokenizer(input_text, return_tensors="pt").to(model.device)

generation_params = {
    "temperature": 0.7,
    "top_k": 50,
    "top_p": 0.9,
    "max_new_tokens": 150,
}

generated_ids = model.generate(**model_inputs, **generation_params)

response = tokenizer.decode(generated_ids[0], skip_special_tokens=True)
print(f"Response:\n{response}")
```
