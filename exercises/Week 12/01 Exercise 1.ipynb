Whenever I read a journalistic article like this, the first thing that comes to mind is the reliability of the information presented. A newspaper article is not the same as a peer-reviewed paper, which must meet rigorous standards to become a scientific article that truly adheres to the scientific method. Therefore, since the source may not be entirely reliable, the opinion I will share below is based on the assumption that the article is truthful.

If DeepMind were indeed acquiring patient data in this manner, it would be deeply concerning, as we are dealing with the private medical histories of millions of individuals. DeepMind does not have doctor-patient confidentiality nor has it sworn any kind of Hippocratic oath. It is a subsidiary of Google, a U.S.-based multinational corporation with thousands of branches worldwide and a strong focus on profit as a publicly traded company. This dynamic inherently prioritizes revenue over individual well-being.

What they will do with the information they have acquired remains unknown, as it is a private company. We also cannot be certain that the data is securely stored, leaving it potentially vulnerable to hackers who might exploit weaknesses in the system. For me, the idea of a non-public entity possessing my personal medical data is completely unacceptable. Companies should adhere to strict standards of integrity and transparency. Although I do not fully agree with the European Union’s current approach to regulating AI and private companies, I believe measures must be taken to ensure that such situations never occur again.
