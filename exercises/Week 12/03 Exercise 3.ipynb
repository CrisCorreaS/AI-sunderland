Before engaging with this paper, I was already well-acquainted with the subject matter, having observed the widespread use of Applicant Tracking Systems (ATS) and other AI-based technologies by companies in both the UK and the EU for the initial screening of resumes. In principle, I have no objections to the use of such technologies, provided that companies adequately train their models. However, when AI systems are trained on unbalanced datasets, they can introduce significant bias against particular demographic groups, as was the case with Amazon’s recruitment algorithm, which exhibited bias against women.

In contemporary recruitment practices, both job seekers and those responsible for screening applications within organizations are increasingly relying on AI. This is an established reality, and I firmly believe that failing to incorporate AI tools places one at a competitive disadvantage. For instance, knowing that most technology companies employ ATS software to filter resumes, candidates are motivated to tailor their CVs to pass these automated screenings. Moreover, with the proliferation of roles generated through AI, job seekers can also leverage AI tools to apply to over 100 positions within a mere 30 minutes.

From an ethical standpoint, I consider the use of AI to be justified, as it is merely another tool as a computer or any mobile device. The ethical concern arises not from the technology itself, but from its misuse, particularly when it is deployed in ways that harm individuals or communities. Provided the AI is well-trained, I see no issue with its application. The more pressing concern, however, is how job applicants can ensure that AI systems are designed with appropriate standards and devoid of bias. In my view, the primary challenge lies in the lack of transparency from both organizations and developers regarding the ethical considerations embedded in these technologies.
