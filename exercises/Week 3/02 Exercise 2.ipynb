## Instructions to complete this problem:
We need a method that needs the coin types we have (e.g: [1, 2, 5, 10, 20, 50, 100, 200]) and our target value (e.g: 93)
1. We need to arrange the coin types in descending order so we can compare first the largest coins
2. We'll create an empty list to store the coins we are going to use
3. We need to check if the first coin type value can fit into the target value amount
4. If the coin value is greater than the target value, we have to subtract the coin value from the target value and add the coin to the result list
5. We repeat the process until the coin type value is lower than the target value
6. We repeat the process with all the coin type values and the remaining target value amount
7. When the target value becomes zero, we end the process and we return the results

```python
def greedy_search(coin_types, target_value):
    coin_types.sort(reverse=True)

    result = []

    for coin_value in coin_types:
        while target_value >= coin_value:
            target_value -= coin_value
            result.append(coin_value)

    return result


coin_types = [1, 2, 5, 10, 20, 50, 100, 200]
target_value_amount = 93

coins_used = greedy_search(coin_types, target_value_amount)

print(f"Coins used to make {target_value_amount}p:", coins_used)
print("Minimum number of coins:", len(coins_used))
```
