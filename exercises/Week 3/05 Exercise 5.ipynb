```python
def greedy_search(coin_types, target_value):
    # Step 1: Sort coin types in descending order
    coin_types.sort(reverse=True)

    # Step 2: Initialize an empty result list
    result = []

    # Step 3: Iterate through each coin type value
    for coin_value in coin_types:
        # Step 4: Use the current coin value as many times as possible
        while target_value >= coin_value:
            target_value -= coin_value  # Subtract the coin value from the target value
            result.append(coin_value)  # Add the coin value to the result list

    # Step 5: Return the result
    return result


# Step 6: Define coin types and the target value amount
coin_types = [1, 2, 5, 10, 20, 50, 100, 200]
target_value_amount = 93

# Step 7: Execute the function and store the result
coins_used = greedy_search(coin_types, target_value_amount)

# Step 8: Print the results
print("Coins used to make 93p:", coins_used)
print("Minimum number of coins:", len(coins_used))
```
