```python
# import required libraries for dataframe and visualization
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import datetime as dt

# import required libraries for clustering
import sklearn
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import linkage
from scipy.cluster.hierarchy import dendrogram
from scipy.cluster.hierarchy import cut_tree
```

```python
# Reading the data on which analysis needs to be done

retail = pd.read_csv(
    "OnlineRetail.csv",
    sep=",",
    encoding="ISO-8859-1",
    header=0,
)
retail.head()
```

```python
# shape of df
retail.shape
```

```python
# df info
retail.info()
```

```python
# df description
retail.describe()
```

```python
# Calculating the Missing Values % contribution in DF
df_null = round(100 * (retail.isnull().sum()) / len(retail), 2)
df_null
```

```python
# Droping rows having missing values
retail = retail.dropna()
retail.shape
```

```python
# Changing the datatype of Customer Id as per Business understanding
retail["CustomerID"] = retail["CustomerID"].astype(str)
```

```python
# New Attribute : Monetary
retail["Amount"] = retail["Quantity"] * retail["UnitPrice"]
rfm_m = retail.groupby("CustomerID")["Amount"].sum()
rfm_m = rfm_m.reset_index()
rfm_m.head()
```

```python
# New Attribute : Frequency
rfm_f = retail.groupby("CustomerID")["InvoiceNo"].count()
rfm_f = rfm_f.reset_index()
rfm_f.columns = ["CustomerID", "Frequency"]
rfm_f.head()
```

```python
# Merging the two dfs
rfm = pd.merge(rfm_m, rfm_f, on="CustomerID", how="inner")
rfm.head()
```

```python
# New Attribute : Recency
# Convert to datetime to proper datatype
retail["InvoiceDate"] = pd.to_datetime(retail["InvoiceDate"], errors="coerce")
```

```python
# Compute the maximum date to know the last transaction date
max_date = max(retail["InvoiceDate"])
max_date
```

```python
# Compute the difference between max date and transaction date
retail["Diff"] = max_date - retail["InvoiceDate"]
retail.head()
```

```python
# Compute last transaction date to get the recency of customers
rfm_p = retail.groupby("CustomerID")["Diff"].min()
rfm_p = rfm_p.reset_index()
rfm_p.head()
```

```python
# Extract number of days only
rfm_p["Diff"] = rfm_p["Diff"].dt.days
rfm_p.head()
```

```python
# Merge tha dataframes to get the final RFM dataframe
rfm = pd.merge(rfm, rfm_p, on="CustomerID", how="inner")
rfm.columns = ["CustomerID", "Amount", "Frequency", "Recency"]
rfm.head()
```

```python
# Outlier Analysis of Amount Frequency and Recency
attributes = ["Amount", "Frequency", "Recency"]
plt.rcParams["figure.figsize"] = [10, 8]
sns.boxplot(
    data=rfm[attributes], orient="v", palette="Set2", whis=1.5, saturation=1, width=0.7
)
plt.title("Outliers Variable Distribution", fontsize=14, fontweight="bold")
plt.ylabel("Range", fontweight="bold")
plt.xlabel("Attributes", fontweight="bold")
```

```python
# Removing (statistical) outliers for Amount
Q1 = rfm.Amount.quantile(0.05)
Q3 = rfm.Amount.quantile(0.95)
IQR = Q3 - Q1
rfm = rfm[(rfm.Amount >= Q1 - 1.5 * IQR) & (rfm.Amount <= Q3 + 1.5 * IQR)]

# Removing (statistical) outliers for Recency
Q1 = rfm.Recency.quantile(0.05)
Q3 = rfm.Recency.quantile(0.95)
IQR = Q3 - Q1
rfm = rfm[(rfm.Recency >= Q1 - 1.5 * IQR) & (rfm.Recency <= Q3 + 1.5 * IQR)]

# Removing (statistical) outliers for Frequency
Q1 = rfm.Frequency.quantile(0.05)
Q3 = rfm.Frequency.quantile(0.95)
IQR = Q3 - Q1
rfm = rfm[(rfm.Frequency >= Q1 - 1.5 * IQR) & (rfm.Frequency <= Q3 + 1.5 * IQR)]
```

```python
# Rescaling the attributes
rfm_df = rfm[["Amount", "Frequency", "Recency"]]

# Instantiate
scaler = StandardScaler()

# fit_transform
rfm_df_scaled = scaler.fit_transform(rfm_df)
rfm_df_scaled.shape
```

```python
rfm_df_scaled = pd.DataFrame(rfm_df_scaled)
rfm_df_scaled.columns = ["Amount", "Frequency", "Recency"]
rfm_df_scaled.head()
```

```python
# k-means with some arbitrary k
kmeans = KMeans(n_clusters=4, max_iter=50)
kmeans.fit(rfm_df_scaled)
```

```python
kmeans.labels_
```

```python
# Elbow-curve/SSD
ssd = []
range_n_clusters = [2, 3, 4, 5, 6, 7, 8]
for num_clusters in range_n_clusters:
    kmeans = KMeans(n_clusters=num_clusters, max_iter=50)
    kmeans.fit(rfm_df_scaled)

    ssd.append(kmeans.inertia_)

# plot the SSDs for each n_clusters
plt.plot(ssd)
```

```python
# Silhouette analysis
range_n_clusters = [2, 3, 4, 5, 6, 7, 8]

for num_clusters in range_n_clusters:

    # intialise kmeans
    kmeans = KMeans(n_clusters=num_clusters, max_iter=50)
    kmeans.fit(rfm_df_scaled)

    cluster_labels = kmeans.labels_

    # silhouette score
    silhouette_avg = silhouette_score(rfm_df_scaled, cluster_labels)
    print(
        "For n_clusters={0}, the silhouette score is {1}".format(
            num_clusters, silhouette_avg
        )
    )
```

```python
# Final model with k=3
kmeans = KMeans(n_clusters=3, max_iter=50)
kmeans.fit(rfm_df_scaled)
```

```python
kmeans.labels_
```

```python
# assign the label
rfm["Cluster_Id"] = kmeans.labels_
rfm.head()
```

```python
# Box plot to visualize Cluster Id vs Frequency
sns.boxplot(x="Cluster_Id", y="Amount", data=rfm)
```

```python
# Box plot to visualize Cluster Id vs Frequency
sns.boxplot(x="Cluster_Id", y="Frequency", data=rfm)
```

```python
# Box plot to visualize Cluster Id vs Recency
sns.boxplot(x="Cluster_Id", y="Recency", data=rfm)
```

```python
# Single linkage:
mergings = linkage(rfm_df_scaled, method="single", metric="euclidean")
dendrogram(mergings)
plt.show()
```

```python
# Complete linkage
mergings = linkage(rfm_df_scaled, method="complete", metric="euclidean")
dendrogram(mergings)
plt.show()
```

```python
# Average linkage
mergings = linkage(rfm_df_scaled, method="average", metric="euclidean")
dendrogram(mergings)
plt.show()
```

```python
# 3 clusters
cluster_labels = cut_tree(mergings, n_clusters=3).reshape(
    -1,
)
cluster_labels
```

```python
# Assign cluster labels
rfm["Cluster_Labels"] = cluster_labels
rfm.head()
```

```python
# Plot Cluster Id vs Amount
sns.boxplot(x="Cluster_Labels", y="Amount", data=rfm)
```

```python
# Plot Cluster Id vs Frequency
sns.boxplot(x="Cluster_Labels", y="Frequency", data=rfm)
```

```python
# Plot Cluster Id vs Recency
sns.boxplot(x="Cluster_Labels", y="Recency", data=rfm)
```
